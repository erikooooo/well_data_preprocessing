```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import rockphypy as rp
```

```python
df = pd.read_csv('../Datasets/DataAVO2.csv')
df.head()
```

```python
def zoeppritz(vp1, vs1, rho1, vp2, vs2, rho2, theta_deg):
    theta = np.deg2rad(theta_deg)
    p = np.sin(theta) / vp1  # ray parameter

    # Snell's law: transmitted angles
    cos1 = np.cos(theta)
    sin2 = p * vp2
    cos2 = np.sqrt(1 - sin2**2)

    # Zoeppritz matrix entries (Aki & Richards style)
    a = rho2 * (1 - 2 * vs2**2 * p**2)
    b = rho1 * (1 - 2 * vs1**2 * p**2)
    c = rho2 * vs2**2
    d = rho1 * vs1**2

    E = b * cos1 + a * cos2
    F = b * sin2 - a * sin1
    G = c * cos1 + d * cos2
    H = c * sin2 - d * sin1

    # Solve linear system for Rpp, Rps, Tpp, Tps
    M = np.array([
        [E, F, -E, -F],
        [G, H, -G, -H],
        [F, -E, F, -E],
        [H, -G, H, -G],
    ])
    rhs = np.array([b * cos1 - a * cos2,
                    d * sin1 - c * sin2,
                    b * sin1 + a * sin2,
                    d * cos1 + c * cos2])

    Rpp, Rps, Tpp, Tps = np.linalg.solve(M, rhs)
    return Rpp, Rps, Tpp, Tps
```

```python
# Clean column names (strip whitespace/newlines)
df.columns = df.columns.str.strip()

# Convert densities to kg/m3
df['rho_shale'] = df['Shale_Vs.1'] * 1000
df['rho_gas']   = df['Gas_p']      * 1000
df['rho_brine'] = df['Brine_p']    * 1000

theta = 30  # incident angle in degrees

rpp_sg, rps_sg = [], []   # Shale -> Gas-sand
rpp_sb, rps_sb = [], []   # Shale -> Brine-sand
rpp_bg, rps_bg = [], []   # Brine-sand -> Gas-sand

for _, row in df.iterrows():
    # Shale–Gas-sand
    Rpp, Rps = rp.AVO.zoeppritz(
        row['Shale_VP'], row['Shale_Vs'], row['rho_shale'],
        row['Gas_Vp'],   row['Gas_Vs'],   row['rho_gas'],
        theta
    )
    rpp_sg.append(Rpp); rps_sg.append(Rps)

    # Shale–Brine-sand
    Rpp, Rps = rp.AVO.zoeppritz(
        row['Shale_VP'],  row['Shale_Vs'],  row['rho_shale'],
        row['Brine_Vp'],  row['Brine_Vs'],  row['rho_brine'],
        theta
    )
    rpp_sb.append(Rpp); rps_sb.append(Rps)

    # Brine-sand–Gas-sand
    Rpp, Rps = rp.AVO.zoeppritz(
        row['Brine_Vp'], row['Brine_Vs'], row['rho_brine'],
        row['Gas_Vp'],   row['Gas_Vs'],   row['rho_gas'],
        theta
    )
    rpp_bg.append(Rpp); rps_bg.append(Rps)

# Save Rpp/Rps to dataframe (used by fluid factor cell below)
df['Rpp_shale_gas']   = rpp_sg;   df['Rps_shale_gas']   = rps_sg
df['Rpp_shale_brine'] = rpp_sb;   df['Rps_shale_brine'] = rps_sb
df['Rpp_brine_gas']   = rpp_bg;   df['Rps_brine_gas']   = rps_bg

# Compute Rp - Rs for each interface
df['Rp_minus_Rs_shale_gas']   = df['Rpp_shale_gas']   - df['Rps_shale_gas']
df['Rp_minus_Rs_shale_brine'] = df['Rpp_shale_brine'] - df['Rps_shale_brine']
df['Rp_minus_Rs_brine_gas']   = df['Rpp_brine_gas']   - df['Rps_brine_gas']

x = df['Index']

plt.figure(figsize=(10, 5))
plt.scatter(x, df['Rp_minus_Rs_shale_brine'], marker='o', s=30, label='Shale / Brine-sand',    color='teal')
plt.scatter(x, df['Rp_minus_Rs_shale_gas'],   marker='^', s=30, label='Shale / Gas-sand',      color='navy')
plt.scatter(x, df['Rp_minus_Rs_brine_gas'],   marker='s', s=30, label='Brine-sand / Gas-sand', color='firebrick')
plt.axhline(0, color='gray', linestyle=':', linewidth=1)
plt.xlabel('Model number', fontsize=12)
plt.ylabel('Rp - Rs', fontsize=12)
plt.title(f'Rp - Rs (θ = {theta}°) per boundary', fontsize=14)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

```python
def fluid_factor(Rp, Rs, Vs, Vp):
    """
    Smith & Gidlow (1987) Fluid Factor:
        F = Rp - 1.16 * (Vs/Vp) * Rs

    F ≈ 0  → brine-sand (on the fluid line)
    F ≠ 0  → gas anomaly (deviation from fluid line)
    """
    return Rp - 1.16 * (Vs / Vp) * Rs


# Use Rpp/Rps already computed and stored in cell above
df['F_shale_gas']   = fluid_factor(
    df['Rpp_shale_gas'],   df['Rps_shale_gas'],
    df['Shale_Vs'], df['Shale_VP']
)
df['F_shale_brine'] = fluid_factor(
    df['Rpp_shale_brine'], df['Rps_shale_brine'],
    df['Shale_Vs'], df['Shale_VP']
)
df['F_brine_gas']   = fluid_factor(
    df['Rpp_shale_gas'],   df['Rps_shale_gas'],
    df['Brine_Vs'], df['Brine_Vp']
)

x = df['Index']

plt.figure(figsize=(10, 5))
plt.scatter(x, df['F_shale_brine'], marker='o', s=30, label='Shale / Brine-sand',    color='teal')
plt.scatter(x, df['F_shale_gas'],   marker='^', s=30, label='Shale / Gas-sand',      color='navy')
plt.scatter(x, df['F_brine_gas'],   marker='s', s=30, label='Brine-sand / Gas-sand', color='firebrick')
plt.axhline(0, color='gray', linestyle=':', linewidth=1)
plt.xlabel('Model number', fontsize=12)
plt.ylabel('Fluid Factor  F', fontsize=12)
plt.title(r'Fluid Factor  $F = R_P - 1.16\,\frac{V_S}{V_P}\,R_S$', fontsize=14)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

```python
# Crossplot: x = Rp - Rs, y = (A + B)/2
# Keep only Brine-sand and Gas-sand series

required_cols = [
    'Rpp_shale_brine', 'Rps_shale_brine',
    'Rpp_shale_gas', 'Rps_shale_gas'
]
missing = [column for column in required_cols if column not in df.columns]
if missing:
    raise ValueError(f"Missing required columns. Run cell 4 first. Missing: {missing}")

# Gas-sand (from shale/gas interface)
x_gas = df['Rpp_shale_gas'] - df['Rps_shale_gas']
y_gas = (df['Rpp_shale_gas'] + df['Rps_shale_gas']) / 2

# Brine-sand (from shale/brine interface)
x_brine = df['Rpp_shale_brine'] - df['Rps_shale_brine']
y_brine = (df['Rpp_shale_brine'] + df['Rps_shale_brine']) / 2

plt.figure(figsize=(10, 6))
plt.scatter(x_brine, y_brine, marker='o', s=35, color='teal', label='Brine-sand')
plt.scatter(x_gas, y_gas, marker='^', s=35, color='navy', label='Gas-sand')

plt.axhline(0, color='gray', linestyle=':', linewidth=1)
plt.axvline(0, color='gray', linestyle=':', linewidth=1)
plt.xlabel('Rp - Rs', fontsize=12)
plt.ylabel('(A + B) / 2', fontsize=12)
plt.title('Crossplot: (A + B)/2 vs (Rp - Rs)', fontsize=14)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

```python
# Plot: y = (A + B)/2 against model number (x = Index)
# Here we use A = Rp and B = Rs for each boundary

required_cols = [
    'Index',
    'Rpp_shale_brine', 'Rps_shale_brine',
    'Rpp_shale_gas', 'Rps_shale_gas',
    'Rpp_brine_gas', 'Rps_brine_gas'
]
missing = [column for column in required_cols if column not in df.columns]
if missing:
    raise ValueError(f"Missing required columns. Run cell 4 first. Missing: {missing}")

# x-axis: model number
x = df['Index']

# y-axis: (A + B)/2 = (Rp + Rs)/2
y_sb = (df['Rpp_shale_brine'] + df['Rps_shale_brine']) / 2
y_sg = (df['Rpp_shale_gas'] + df['Rps_shale_gas']) / 2
y_bg = (df['Rpp_brine_gas'] + df['Rps_brine_gas']) / 2

plt.figure(figsize=(10, 6))
plt.scatter(x, y_sb, marker='o', s=35, color='teal', label='Shale / Brine-sand')
plt.scatter(x, y_sg, marker='^', s=35, color='navy', label='Shale / Gas-sand')
plt.scatter(x, y_bg, marker='s', s=35, color='firebrick', label='Brine-sand / Gas-sand')

plt.axhline(0, color='gray', linestyle=':', linewidth=1)
plt.xlabel('Model number', fontsize=12)
plt.ylabel('(A + B) / 2', fontsize=12)
plt.title('Crossplot: (A + B)/2 vs Model number', fontsize=14)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```
